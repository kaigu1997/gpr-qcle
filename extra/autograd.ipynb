```python
import torch
from IPython.display import display
torch.set_default_device("cuda")
torch.set_default_dtype(torch.float64)
```

```python
x = torch.randn(3, 2, requires_grad=True)
x_ind = torch.randn(4, 2, requires_grad=True)
x_all = torch.randn(5, 2, requires_grad=True)
y_all = torch.randn(5, requires_grad=True)
l = torch.rand(2, requires_grad=True)

out = ((x[:, None] - x_ind[None]) / l).sum(-1) @ ((x_ind[:, None] - x_all[None]) / l).sum(-1) @ y_all

display(
	(((x_ind[:, None] - x_all[None]) / l).sum(-1) @ y_all).sum() / l,
	torch.autograd.grad(out, x, torch.ones_like(out), True, False, True, True, False, True)[0].detach(),
	torch.autograd.grad(out, x_ind, torch.eye(out.numel()), True, False, True, True, True, True)[0].detach().reshape(out.shape + x_ind.shape),
	torch.autograd.grad(out, x_all, torch.eye(out.numel()), True, False, True, True, True, True)[0].detach().reshape(out.shape + x_all.shape),
	torch.autograd.grad(out, y_all, torch.eye(out.numel()), True, False, True, True, True, True)[0].detach().reshape(out.shape + y_all.shape),
	torch.autograd.grad(out, l, torch.eye(out.numel()), True, False, True, True, True, True)[0].detach().reshape(out.shape + l.shape),
	-((x[:, None] - x_ind[None]) / l**2).sum(-1) @ ((x_ind[:, None] - x_all[None]) / l).sum(-1) @ y_all - ((x[:, None] - x_ind[None]) / l).sum(-1) @ ((x_ind[:, None] - x_all[None]) / l**2).sum(-1) @ y_all
)
```
