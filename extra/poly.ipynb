```python
import torch

torch.manual_seed(0)
torch.set_default_dtype(torch.double)
```

```python
n = int(torch.randint(10, 20, (1,)).item())
n
```

```python
coe = torch.randint(0, 10, (n,))
values = torch.tensor([[10, 5], [2, 1]])
print(coe, values, sep="\n")
```

```python
print("exact value:")
exact = sum((coe[i] * values ** i for i in range(n)), start=torch.zeros_like(values))
print(exact)
```

```python
print("Horner's scheme:")
result = torch.full_like(values, coe[-1].item())
for i in range(n - 2, -1, -1):
	result = values * result + coe[i]
print(result, torch.abs(result - exact).sum().item(), sep="\n")
```

```python
print("Estrin's scheme:")
result = coe[:, *([torch.newaxis] * values.ndim)] * torch.ones_like(values)
n = coe.shape[0] - 1
pow = values
while n > 0:
	print(result[:, 0, 0], n)
	for i in range(0, n, 2):
		result[i//2] = result[i] + result[i+1] * pow
	if n % 2 == 0:
		result[n//2] = result[n]
	pow = pow * pow
	n //= 2
print(result[0], torch.abs(result[0] - exact).sum().item(), sep="\n")
```
